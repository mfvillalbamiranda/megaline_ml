# 1. Abrir y examinar el archivo de datos

```python
# importación de librerías
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

import matplotlib.pyplot as plt
```

```python
# cargar los datos
df = pd.read_csv('users_behavior.csv')
df.info()
```

## Revisión de calidad de los datos

```python
# verificar si hay presencia de valores nulos

print(df.head(10))
print("Valores nulos en columna calls:", df['calls'].isnull().sum())
print("Valores nulos en columna messages:", df['messages'].isnull().sum())
```

```python
# verificar valores ducplicados

print("número de valores duplicados: ", df.duplicated().sum())
```

```python
# cambiar las llamadas y mensajes a enteros
df['calls'] = df['calls'].astype(int)
df['messages'] = df['messages'].astype(int)

df.info()
```

# 2. Segmentación de los datos

```python
# Segmentación de los datos
features = df.drop(columns='is_ultra')
target = df['is_ultra']

print("tamaño de las características:", features.shape)
print("tamaño del objetivo:", target.shape)
```

```python
# dividir el conjunto de datos en 60% entrenamiento, 20% validación, 20% prueba

features_temp, features_valid, target_temp, target_valid  = train_test_split(features, target, test_size=0.20, random_state=12345) 

features_train, features_test, target_train, target_test  = train_test_split(features_temp, target_temp, test_size=0.25, random_state=12345) #toca multiplicar por 0.25 para obtener el 20% del total

# Verificar las formas de los conjuntos de datos

print("Características de entrenamiento:", features_train.shape, "objetivos de entrenamiento:", target_train.shape)
print("Características de validación:", features_valid.shape, "objetivos de validación:", target_valid.shape)
print("Características de prueba:", features_test.shape, "objetivos de prueba:", target_test.shape)

print(f'Porcentaje entrenamiento: {round(len(target_train) * 100 / len(target))}%')
print(f'Porcentaje validación: {round(len(target_valid) * 100 / len(target))}%')
print(f'Porcentaje prueba: {round(len(target_test) * 100 / len(target))}%')
```

# 3. Investigar la calidad de los diferentes modelos

## Árbol de decisión

```python
# definición del modelo y características
tree_model = DecisionTreeClassifier(random_state=12345, max_depth=10)

# entrenamiento del modelo
tree_model.fit(features_train, target_train) 

# predicción en el conjunto de validación
tree_predictions = tree_model.predict(features_valid)

# cálculo de exactitud
tree_accuracy = accuracy_score(target_valid, tree_predictions) 

print("Exactitud del modelo de árbol de decisión:", tree_accuracy)
```

### Evaluación del sobreajuste

```python
max_depths = range(1, 50)
train_scores, test_scores = [], []
for d in max_depths:
    clf = DecisionTreeClassifier(max_depth=d, random_state=12345).fit(features_train, target_train)
    train_scores.append(accuracy_score(target_train, clf.predict(features_train)))
    test_scores.append(accuracy_score(target_test, clf.predict(features_test)))
plt.plot(max_depths, train_scores, label="Train")
plt.plot(max_depths, test_scores, label="Test")
plt.xlabel("max_depth")
plt.ylabel("Accuracy")
plt.title("Subajuste vs Sobreajuste")
plt.legend()
plt.show()
```

## Bosque aleatorio

```python
# definición del modelo y características
forest_model = RandomForestClassifier(random_state=12345, n_estimators=40, max_depth=10) 

# entrenamiento del modelo
forest_model.fit(features_train, target_train)

# predicción en el conjunto de validación
forest_predictions = forest_model.predict(features_valid)

# cálculo de exactitud
forest_accuracy = accuracy_score(target_valid, forest_predictions) 

print("Exactitud del modelo de bosque aleatorio:", forest_accuracy)
```

### Evaluación del sobreajuste

```python
max_estimator = range(1, 50)
train_scores, test_scores = [], []
for d in max_estimator:
    clf = RandomForestClassifier(random_state=12345, n_estimators=d, max_depth=10).fit(features_train, target_train)
    train_scores.append(accuracy_score(target_train, clf.predict(features_train)))
    test_scores.append(accuracy_score(target_test, clf.predict(features_test)))
plt.plot(max_depths, train_scores, label="Train")
plt.plot(max_depths, test_scores, label="Test")
plt.xlabel("max_estimators")
plt.ylabel("Accuracy")
plt.title("Subajuste vs Sobreajuste")
plt.legend()
plt.show()
```

## Regresión logística

```python
# definición del modelo y características
regression_model = LogisticRegression(random_state=12345, solver='liblinear') 

# entrenamiento del modelo
regression_model.fit(features_train, target_train)

# predicción en el conjunto de validación
regression_predictions = regression_model.predict(features_valid)

# cálculo de exactitud
regression_accuracy = accuracy_score(target_valid, regression_predictions) 

print("Exactitud del modelo de regresión logística:", regression_accuracy)
```

## Descripción de los hallazgos

Luego de evaluar los tres modelos de clasificación, se llega a la conclusión que el que tuvo una mejor exactitud es el de bosque aleatorio, por lo que hace un paso adicional para comprobar estos resultados con los datos de prueba para corroborar

# 4. Comprobar la calidad del modelo usando el conjunto de prueba

```python
# predicción árbol de decisión
tree_test_predictions = tree_model.predict(features_test)

# predicción bosque aleatorio
forest_test_predictions = forest_model.predict(features_test)

# predicción regresión logística
reg_test_predictions = regression_model.predict(features_test)

print("Calidad del modelo árbol de decisión usando el conjunto de prueba:", accuracy_score(target_test, tree_test_predictions))
print("Calidad del modelo bosque aleatorio usando el conjunto de prueba:", accuracy_score(target_test, forest_test_predictions))
print("Calidad del modelo regresión lineal usando el conjunto de prueba:", accuracy_score(target_test,reg_test_predictions))
```

## Se confirma con los datos de prueba que el mejor modelo sigue siendo el bosque aleatorio
